```python
import pandas as pd
catboost_test = pd.read_csv('outputs/predictions/catboost_test_predictions.csv')
xgboost_test = pd.read_csv('outputs/predictions/xgboost_test_predictions.csv')
survival_test = pd.read_csv('outputs/predictions/survival_logit_test_predictions.csv')
transformer_test = pd.read_csv('outputs/predictions/temporal_transformer_test_predictions.csv')
logistic_test = pd.read_csv('outputs/predictions/logistic_test_predictions.csv')

```

```python
catboost_test = catboost_test.drop_duplicates(subset='id_student')
survival_test = survival_test.drop_duplicates(subset='id_student')
xgboost_test = xgboost_test.drop_duplicates(subset='id_student')
logistic_test = logistic_test.drop_duplicates(subset='id_student')
transformer_test = transformer_test.drop_duplicates(subset='id_student')
```

```python
print("Cat",len(catboost_test))
print("Survival",len(survival_test))
print("XG",len(xgboost_test))
print("Logistic",len(logistic_test))
print("Transformer",len(transformer_test))
```

```python
catboost_test.to_csv('outputs/predictions/catboost_test_predictions.csv',index=False)
xgboost_test.to_csv('outputs/predictions/xgboost_test_predictions.csv',index=False)
survival_test.to_csv('outputs/predictions/survival_logit_test_predictions.csv',index=False)
transformer_test.to_csv('outputs/predictions/temporal_transformer_test_predictions.csv',index=False)
logistic_test.to_csv('outputs/predictions/logistic_test_predictions.csv',index=False)
```

```python
table_list = [catboost_test,xgboost_test,logistic_test,survival_test,transformer_test]
common_id = set.intersection(*[set(t["id_student"]) for t in table_list])
#Each dataframe's student ID's are kept in a set. From these five sets, the common ID's are kept in common_id.
print(len(common_id))
```

```python
xgboost_test = xgboost_test[xgboost_test['id_student'].isin(common_id)]
logistic_test = logistic_test[logistic_test['id_student'].isin(common_id)]
survival_test = survival_test[survival_test['id_student'].isin(common_id)]
transformer_test = transformer_test[transformer_test['id_student'].isin(common_id)]
```

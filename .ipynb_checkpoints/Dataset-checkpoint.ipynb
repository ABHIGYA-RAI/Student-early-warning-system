```python
#code_module = The code of the course
#course_presentation = 2013J, the course started in 2013 and J means October as B is January
#imd_band = How deprived or poor the student's home area is. 0-10% is the most deprived whereas 90-100% is the least deprived.
#num_of_prev_attempts = How many times has the student taken this course before
#studied_credits = It measures the workload of the student. More credits means they have taken multiple courses. A student is likely to withdraw from a course if they have heavier workload.
```

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

CUTOFF_DAY = 28   # prediction point: end of week 4

# Load and join
info = pd.read_csv("studentInfo.csv")
reg = pd.read_csv("studentRegistration.csv")
df = info.merge(reg, on=["code_module", "code_presentation", "id_student"], how="left")
df["dropout"] = (df["final_result"] == "Withdrawn").astype(int)

# Clean imd_band
df["imd_band"] = df["imd_band"].str.replace("%", "", regex=False).fillna("Unknown")
df.loc[df["imd_band"] != "Unknown", "imd_band"] += "%"

# Keep only students still enrolled at the prediction point
n_all = len(df)
still_in = df["date_unregistration"].isna() | (df["date_unregistration"] > CUTOFF_DAY)
registered = df["date_registration"].isna() | (df["date_registration"] <= CUTOFF_DAY)
df = df[still_in & registered].copy()
print(f"All rows: {n_all} | eligible: {len(df)} | removed: {n_all - len(df)}")
print("Dropout rate:", round(df["dropout"].mean(), 3))

# Model inputs (protected attributes excluded on purpose)
cat_cols = ["code_module", "region", "highest_education", "imd_band"]
num_cols = ["num_of_prev_attempts", "studied_credits"]

# Chronological split: train on earlier runs, test on the latest
train = df["code_presentation"].isin(["2013B", "2013J", "2014B"])
test = df["code_presentation"] == "2014J"
X_train, y_train = df.loc[train, cat_cols + num_cols], df.loc[train, "dropout"]
X_test, y_test = df.loc[test, cat_cols + num_cols], df.loc[test, "dropout"]

pre = ColumnTransformer([
    ("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols),
    ("num", StandardScaler(), num_cols)])
model = Pipeline([("prep", pre),
                  ("clf", LogisticRegression(max_iter=1000, class_weight="balanced"))])
model.fit(X_train, y_train)

proba = model.predict_proba(X_test)[:, 1]
print("Train rows:", len(X_train), "| Test rows:", len(X_test))
print("Sanity ROC-AUC:", round(roc_auc_score(y_test, proba), 3))

# Save for the next steps (protected attributes kept here for auditing only)
out = df.loc[test, ["id_student", "code_module", "code_presentation",
                    "gender", "disability", "age_band", "imd_band", "dropout"]].copy()
out["prob_baseline"] = proba
out.to_csv("baseline_predictions.csv", index=False)
df.to_csv("cohort_day28.csv", index=False)
print("Saved baseline_predictions.csv and cohort_day28.csv")
```
